```python
import matplotlib.pyplot as plt
import numpy as np

# Data for Enceladus
depth_enceladus = np.array([0, 50, 100])  # Example depths in km
temperature_enceladus = np.array([245, 270, 275])  # Example temperatures in K

# Data for Europa
depth_europa = np.array([0, 500, 1000, 1500])  # Example depths in km
temperature_europa = np.array([200, 240, 260, 273])  # Example temperatures in K

# Plot HR diagram for Enceladus
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(temperature_enceladus, depth_enceladus, color='cyan', linestyle='--', marker='o', label='Enceladus')
plt.gca().invert_yaxis()  # Invert y-axis to represent depth correctly
plt.title("Enceladus HR Diagram")
plt.xlabel("Temperature (K)")
plt.ylabel("Depth (km)")
plt.legend()
plt.grid(True)

# Plot HR diagram for Europa
plt.subplot(1, 2, 2)
plt.plot(temperature_europa, depth_europa, color='magenta', linestyle='-', marker='s', label='Europa')
plt.gca().invert_yaxis()  # Invert y-axis to represent depth correctly
plt.title("Europa HR Diagram")
plt.xlabel("Temperature (K)")
plt.ylabel("Depth (km)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Example Data for Enceladus (replace with actual data if available)
depth_enceladus = np.array([0, 10, 50, 100])  # Depth in km
temperature_enceladus = np.array([245, 260, 270, 273])  # Temperature in K

# Example Data for Europa (replace with actual data if available)
depth_europa = np.array([0, 500, 1000, 1500, 2000])  # Depth in km
temperature_europa = np.array([200, 230, 250, 270, 273])  # Temperature in K

# Constants
melting_temperature_water = 273  # K (melting point of ice)
ice_transition_depth_enceladus = 50  # Example depth of ice-water transition in km
ice_transition_depth_europa = 1500  # Example depth of ice-water transition in km

# Create the plots
plt.figure(figsize=(12, 6))

# Enceladus HR Diagram
plt.subplot(1, 2, 1)
plt.plot(temperature_enceladus, depth_enceladus, color='cyan', linestyle='--', marker='o', label='Temperature Profile')
plt.axvline(melting_temperature_water, color='red', linestyle=':', label='Melting Point of Ice')
plt.axhline(ice_transition_depth_enceladus, color='blue', linestyle='--', label='Ice-Water Transition')
plt.gca().invert_yaxis()  # Depth increases downward
plt.title("Enceladus HR Diagram")
plt.xlabel("Temperature (K)")
plt.ylabel("Depth (km)")
plt.grid(True)
plt.legend()
plt.annotate("Surface Ice", xy=(245, 5), xytext=(250, 20), 
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.annotate("Subsurface Ocean", xy=(273, 55), xytext=(250, 70), 
             arrowprops=dict(facecolor='black', arrowstyle='->'))

# Europa HR Diagram
plt.subplot(1, 2, 2)
plt.plot(temperature_europa, depth_europa, color='magenta', linestyle='-', marker='s', label='Temperature Profile')
plt.axvline(melting_temperature_water, color='red', linestyle=':', label='Melting Point of Ice')
plt.axhline(ice_transition_depth_europa, color='blue', linestyle='--', label='Ice-Water Transition')
plt.gca().invert_yaxis()  # Depth increases downward
plt.title("Europa HR Diagram")
plt.xlabel("Temperature (K)")
plt.ylabel("Depth (km)")
plt.grid(True)
plt.legend()
plt.annotate("Thick Ice Layer", xy=(230, 700), xytext=(240, 1000), 
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.annotate("Subsurface Ocean", xy=(273, 1600), xytext=(250, 1800), 
             arrowprops=dict(facecolor='black', arrowstyle='->'))

# Finalize layout
plt.tight_layout()
plt.show()
```
