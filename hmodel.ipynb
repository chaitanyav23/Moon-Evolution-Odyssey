```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Load the dataset
file_path = '/content/icy_moons_dataset.csv'
data = pd.read_csv(file_path)

# Display basic information about the dataset
data.info(), data.head()
```

```python
# Extract rows where ESI_Global is not null for training the Decision Tree model
known_esi_data = data[data['ESI_Global'].notnull()]
unknown_esi_data = data[data['ESI_Global'].isnull()]

# Features for imputation
features = [
    'Radius_km', 'Surface_Temp_K', 'Albedo', 'Surface_Gravity_m_s2',
    'Escape_Velocity_km_s', 'Radiation_mSv_day', 'Tidal_Heating_GW'
]

# Splitting data into features (X) and target (y)
X_known = known_esi_data[features]
y_known = known_esi_data['ESI_Global']

# Filling missing feature values with the column mean (if any)
X_known = X_known.fillna(X_known.mean())
X_unknown = unknown_esi_data[features].fillna(X_known.mean())

# Train Decision Tree Regressor for imputation
decision_tree = DecisionTreeRegressor(random_state=42)
decision_tree.fit(X_known, y_known)

# Predict missing ESI_Global values
imputed_values = decision_tree.predict(X_unknown)

# Add imputed values as a new column
data['Imputed_ESI_Global'] = data['ESI_Global']
data.loc[data['ESI_Global'].isnull(), 'Imputed_ESI_Global'] = imputed_values

# Display a sample of the updated dataset
data[['Moon', 'ESI_Global', 'Imputed_ESI_Global']].head(10)
```

```python
# Use the imputed column for training the Random Forest model
X_full = data[features].fillna(X_known.mean())
y_full = data['Imputed_ESI_Global']

# Train-test split for evaluation
X_train, X_test, y_train, y_test = train_test_split(X_full, y_full, test_size=0.3, random_state=42)

# Train the Random Forest Regressor
random_forest = RandomForestRegressor(random_state=42, n_estimators=100)
random_forest.fit(X_train, y_train)

# Predict on the test set and evaluate
y_pred = random_forest.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# Predict missing ESI_Global values using the trained Random Forest model
rf_predictions = random_forest.predict(X_full[data['ESI_Global'].isnull()])

mse, r2, rf_predictions
```
